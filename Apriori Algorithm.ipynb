```python
# ============================================================
# APRIORI MARKET BASKET ANALYSIS
# Retail Market Basket Transactions Dataset
# ============================================================

# ------------------------------------------------------------
# 0. HIDE UNNECESSARY WARNINGS
# ------------------------------------------------------------

import warnings

warnings.filterwarnings(
    "ignore",
    category=DeprecationWarning
)


# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.frequent_patterns import (
    apriori,
    association_rules
)


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(
    "Market_Basket_Optimisation.csv",
    header=None
)

print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 3. CREATE TRANSACTIONS
# ============================================================

transactions = []

for i in range(len(df)):

    transaction = []

    for item in df.iloc[i]:

        if pd.notna(item):

            item = str(item).strip()

            if item != "":

                transaction.append(item)

    transactions.append(transaction)


print("\nTotal Transactions:")
print(len(transactions))


# ============================================================
# 4. FIND UNIQUE ITEMS
# ============================================================

unique_items = set()

for transaction in transactions:

    for item in transaction:

        unique_items.add(item)


unique_items = sorted(unique_items)


print("\nTotal Unique Items:")
print(len(unique_items))


# ============================================================
# 5. CREATE ONE-HOT ENCODED DATASET
# ============================================================

basket = pd.DataFrame(
    False,
    index=range(len(transactions)),
    columns=unique_items
)


for i, transaction in enumerate(transactions):

    for item in transaction:

        basket.loc[i, item] = True


print("\nOne-Hot Encoded Dataset:")
print(basket.head())


# ============================================================
# 6. APPLY APRIORI ALGORITHM
# ============================================================

print("\n" + "=" * 70)
print("APRIORI ALGORITHM")
print("=" * 70)

frequent_itemsets = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)


print("\nNumber of Frequent Itemsets:")
print(len(frequent_itemsets))


# ============================================================
# 7. DISPLAY FREQUENT ITEMSETS
# ============================================================

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)


print("\nTop 20 Frequent Itemsets:")

print(
    frequent_itemsets.head(20)
)


# ============================================================
# 8. CREATE ASSOCIATION RULES
# ============================================================

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.20
)


print("\n" + "=" * 70)
print("ASSOCIATION RULES")
print("=" * 70)

print("\nNumber of Association Rules:")
print(len(rules))

# Display top N rules
print("\nTop 10 Association Rules:")
print(
    rules.sort_values(
        by='confidence',
        ascending=False
    ).head(10)
)

# ============================================================
# 9. VISUALIZE ASSOCIATION RULES
# ============================================================

print("\n" + "=" * 70)
print("VISUALIZATION")
print("=" * 70)

if not rules.empty:
    plt.figure(figsize=(10, 6))
    plt.scatter(rules["support"], rules["confidence"], alpha=0.5)
    plt.xlabel("Support")
    plt.ylabel("Confidence")
    plt.title("Association Rules: Support vs Confidence")
    plt.grid(True)
    plt.show()
else:
    print("No association rules found to visualize with the given parameters.")

# ============================================================
# 10. SAVE RESULTS
# ============================================================

print("\n" + "=" * 70)
print("SAVE RESULTS")
print("=" * 70)

# Save frequent itemsets
frequent_itemsets.to_csv("apriori_frequent_itemsets.csv", index=False)
print("Frequent itemsets saved to apriori_frequent_itemsets.csv")

# Save association rules
rules.to_csv("apriori_association_rules.csv", index=False)
print("Association rules saved to apriori_association_rules.csv")

print("\nApriori analysis completed successfully!")
```

### Itemset Length Distribution

Let's examine the distribution of itemset lengths to understand how many items typically form a frequent itemset.

```python
# ============================================================
# 11. ITEMSET LENGTH DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("ITEMSET LENGTH DISTRIBUTION")
print("=" * 70)

# Calculate the length of each itemset
frequent_itemsets['itemset_length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))

# Count the occurrences of each itemset length
itemset_length_counts = frequent_itemsets['itemset_length'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
itemset_length_counts.plot(kind='bar')
plt.title('Distribution of Frequent Itemset Lengths')
plt.xlabel('Itemset Length')
plt.ylabel('Number of Frequent Itemsets')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
